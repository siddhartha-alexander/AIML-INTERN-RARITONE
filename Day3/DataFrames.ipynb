```python
import pandas as pd

data={
    'Name':['Siddhartha','Rahul','Charan','Chandu'],
    'Age':[18,19,20,21],
    'height':[5.9,5.4,5.9,5]
}

df=pd.DataFrame(data)
print(df)
```

```python
### New column
df['job']=['Data Sceintist','Doctor','Doctor','Police']
print(df)
```

```python
### adding a new row
new_row=pd.DataFrame([
    {'Name':'Walt','Age':50,'height':5.8,'job':'chemist'}
])

df=pd.concat([df,new_row])
```

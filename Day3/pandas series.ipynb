### SERIES

series is a 1 Dimensional labelled array that can hold any data type 
think of it like single column in a spread sheet

```python
import pandas as pd

data=[100,101,102]
series=pd.Series(data)
print(series)
```

```python
data=[100.1,101.2,102.3]
series=pd.Series(data)
print(series)
```

```python
data=['A','B','C']
series=pd.Series(data)
print(series)
```

```python
data=[True,False]
series=pd.Series(data)
print(series)
```

```python
data=[100,101,102]
series=pd.Series(data,index=['a','b','c'])
print(series)
print(series.loc['a'])
```

```python
data=[100,101,102]
series=pd.Series(data)
print(series.iloc[0])
print(series.iloc[1])
print(series.iloc[2])

```

```python
data=[100,101,102]
series=pd.Series(data,index=['a','b','c'])
print(series[series>100])
```

### 1.ARTIFICIAL INTELLIGENCE 

Artificial Intelligence (AI) is a technology that lets computer and machines mimic human learning ,reasoning and problem solving 

### 2.MACHINE LEARNING 

Machine Learning is a subset of Artificial Intelligence ,where machine learns pattern from data make predictions/decisions without being explicitly programmed for every rule

it is further classified into 3 types they are :

        1.Supervised Learning

        2.Unsupervised Learning 

        3.Reinforcement Learning 

### Supervised Learning

Supervised Learning uses labeled data and known answers to train models for precise predictions

supervised learning further classified into two types:

        1.Regression
            
            a.Linear Regression

        2.Classification

            a.Decision Tree

            b.Logistic Regression

            c.Random Forest

### UNSUPERVISED LEARNING 

Unsupervised learning uses raw and unlabeled data so that algorithm can independently discover hidden patterns and groupings    

Unsupervised Learning furthe classified into 2 types:

        1.Clustering

        2.Association 

### REINFORCEMENT LEARNING

Reinforcement Learning is a machine learning method where an AI agent learns to make decisions by interacting with Environment

### 3.DEEP LEARNING 

Deep learning is a subset of Machine Learning inspired by the structure and function of the human brain

It uses multilayered artificial neural network to automatically learns pattern from massive amount of data

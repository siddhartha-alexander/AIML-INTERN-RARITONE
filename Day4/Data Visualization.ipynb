### DATA VISUALIZATION

Data Visualization in python is the practice of converting raw datasets into graphical representations such a charts,graphs ,maps and dashboards using the python programming language

```python
import matplotlib.pyplot as plt
```

```python
x=[2018,2019,2020,2021,2022,2023]
y=[2,3,4,5,6,10]
plt.plot(x,y)
plt.show
```

```python
import matplotlib.pyplot as plt
import numpy as np

x=np.array([100,200,300,400])
y=np.array([5,10,15,20])
plt.plot(x,y)
plt.show
```

```python
import matplotlib.pyplot as plt
import numpy as np

x=np.array([100,200,300,400])
y=np.array([5,10,15,20])
plt.plot(x,y,marker='o',markersize=20,markerfacecolor='red',markeredgecolor='red',linestyle='dashed',
         linewidth=4,color='blue')
plt.show
```

```python
x=np.array([100,200,300,400])
y1=np.array([5,10,15,20])
y2=np.array([4,9,12,4])
line_style=dict(marker='o',markersize=20,markerfacecolor='red',markeredgecolor='red',linestyle='dashed',
         linewidth=4,color='blue')
plt.plot(x,y,**line_style)
plt.plot(x,y2,**line_style)
plt.show
```

```python
x=np.array([100,200,300,400])
y1=np.array([5,10,18,27])
y2=np.array([4,9,16,28])

plt.plot(x,y)
plt.plot(x,y2)

plt.title('Title',family='Arial',fontweight='bold',color='pink')
plt.xlabel('x_axis',family='Arial',fontweight='bold',color='purple')
plt.ylabel('y_axis',family='Arial',fontweight='bold',color='purple')
plt.show
```

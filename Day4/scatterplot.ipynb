### SCATTER PLOT 

Scatter Plot show relationship between two variables ,helps to identify correlation

```python
import matplotlib.pyplot as plt
import numpy as np
x=[0,1,1,2,2,3,1,3,4,5]
grades=[55,60,70,68,59,44,73,75,66,59]

plt.scatter(x,grades)
plt.show()
```

### HISTOGRAM

Histogram is a visual representation of the distribution of quantitative data 
They group values into bins 
and count how many falls into each range

```python
import matplotlib.pyplot as plt
import numpy as np

scores=np.random.normal(loc=80,scale=10,size=100) #loc will set middle value to 80 and scale is deviation 

plt.hist(scores)
plt.show()
```

```python
scores=np.random.normal(loc=80,scale=10,size=100)
scores=np.clip(scores,0,100)
plt.hist(scores)
plt.show()
```

```python
scores=np.random.normal(loc=80,scale=10,size=100)
scores=np.clip(scores,0,100)
plt.hist(scores,bins=5,color='lightgreen',edgecolor='black')
plt.title('Exam Scores')
plt.xlabel('Score')
plt.ylabel('no.of students')
plt.show()
```

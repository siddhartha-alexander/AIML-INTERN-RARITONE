### BAR CHART  

Bar Chart compares categories of data by representing each category with a bar

```python
import matplotlib.pyplot as plt
import numpy as np

categories=['apple','banana','jackfruit','orange','guava']
count=[7,4,6,2,9]

plt.bar(categories,count)
plt.title('Daily consumption')
plt.xlabel('fruits')
plt.ylabel('consumtion')
plt.grid()
plt.show()
```

```python
categories=['apple','banana','jackfruit','orange','guava']
count=[7,4,6,2,9]

plt.barh(categories,count)
plt.title('Daily consumption')
plt.xlabel('fruits')
plt.ylabel('consumtion')
plt.show()
```

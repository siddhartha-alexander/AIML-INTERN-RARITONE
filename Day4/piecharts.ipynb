### PIE CHARTS

Pie chart is a circular chart that divided into slices to show percentages of the total

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories=['Freshmen','Sophmores','Juniors','Seniors']
values=np.array([150,173,164,228])
plt.pie(values)
plt.show()
```

```python
categories=['Freshmen','Sophmores','Juniors','Seniors']
values=np.array([150,173,164,228])
plt.pie(values,labels=categories,autopct='%1.1f')
plt.title('College Code')
plt.show()
```

```python
categories=['Freshmen','Sophmores','Juniors','Seniors']
values=np.array([150,173,164,228])
plt.pie(values,labels=categories,autopct='%1.1f',explode=[0,0,0,0.1],shadow=True,startangle=180)
plt.title('College Code')
plt.show()
```
